```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- PARAMETERS ----------
years = 40
conflict_year = 10

alpha = 0.33          # capital share
delta = 0.05          # depreciation rate
shock_size = 0.4      # 40% capital destruction

# Savings scenarios
s_low = 0.15
s_medium = 0.25
s_high = 0.35


# ---------- SOLOW SIMULATION FUNCTION ----------
def simulate_solow(savings_rate):

    K = [100]   # initial capital stock
    Y = []

    for t in range(years):

        # Production function
        output = K[-1] ** alpha
        Y.append(output)

        # War shock
        if t == conflict_year:
            K.append(K[-1] * (1 - shock_size))
            continue

        # Capital accumulation
        new_K = savings_rate * output + (1 - delta) * K[-1]
        K.append(new_K)

    return Y


# ---------- RUN SCENARIOS ----------
Y_low = simulate_solow(s_low)
Y_medium = simulate_solow(s_medium)
Y_high = simulate_solow(s_high)


# ---------- PLOT ----------
plt.figure(figsize=(10,6))

plt.plot(Y_low, label="Low Savings (15%)")
plt.plot(Y_medium, label="Medium Savings (25%)")
plt.plot(Y_high, label="High Savings (35%)")

plt.axvline(conflict_year, linestyle="--", alpha=0.6)

plt.title("Solow Model with War Shock")
plt.xlabel("Year")
plt.ylabel("Output (GDP)")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------- PARAMETERS ----------
years = 40
conflict_year = 10

alpha = 0.33          # capital share
delta = 0.05          # depreciation rate
shock_size = 0.4      # 40% capital destruction

# Savings scenarios
s_low = 0.25
s_medium = 0.35
s_high = 0.45


# ---------- SOLOW SIMULATION FUNCTION ----------
def steady_state_k(s):
    return (s / delta) ** (1 / (1 - alpha))


def simulate_solow(savings_rate):

    K_star = steady_state_k(savings_rate)
    K = [K_star]   # start at steady state
    Y = []

    for t in range(years):

        output = K[-1] ** alpha
        Y.append(output)

        if t == conflict_year:
            K.append(K[-1] * (1 - shock_size))
            continue

        new_K = savings_rate * output + (1 - delta) * K[-1]
        K.append(new_K)

    return Y


# ---------- RUN SCENARIOS ----------
Y_low = simulate_solow(s_low)
Y_medium = simulate_solow(s_medium)
Y_high = simulate_solow(s_high)


# ---------- PLOT ----------
plt.figure(figsize=(10,6))

plt.plot(Y_low, label="Low Savings (15%)")
plt.plot(Y_medium, label="Medium Savings (25%)")
plt.plot(Y_high, label="High Savings (35%)")

plt.axvline(conflict_year, linestyle="--", alpha=0.6)

plt.title("Solow Model with War Shock")
plt.xlabel("Year")
plt.ylabel("Output (GDP)")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("Steady state K (low):", steady_state_k(s_low))
print("Steady state K (medium):", steady_state_k(s_medium))
print("Steady state K (high):", steady_state_k(s_high))

```
