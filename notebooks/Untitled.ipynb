```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("rwanda_gdp_growth.csv", skiprows=4)

# Inspect columns (important for World Bank files)
print(df.columns)

# Keep only year columns
df = df.melt(id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"], 
             var_name="Year", 
             value_name="GDP Growth")

# Convert Year to numeric
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

# Drop missing values
df = df.dropna(subset=["GDP Growth"])

# Plot
plt.figure()
plt.plot(df["Year"], df["GDP Growth"])
plt.title("GDP Growth (Annual %) - Rwanda")
plt.xlabel("Year")
plt.ylabel("GDP Growth %")
plt.show()
```
