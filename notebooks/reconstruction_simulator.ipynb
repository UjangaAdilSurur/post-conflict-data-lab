```python
import numpy as np
import matplotlib.pyplot as plt


years = 30
conflict_year = 5
baseline_growth = 0.03
shock_size = -0.25


reconstruction_low = 0.04
reconstruction_medium = 0.06
reconstruction_high = 0.08



def simulate_recovery(reconstruction_rate):

    gdp = [100]  

    for year in range(1, years):

        if year == conflict_year:
            growth = shock_size

        elif year > conflict_year:

            years_since_conflict = year - conflict_year

            
            decay_factor = np.exp(-0.2 * years_since_conflict)
            reconstruction_component = (
                reconstruction_rate - baseline_growth
            ) * decay_factor

            
            volatility = 0.05 * np.exp(-0.3 * years_since_conflict)
            random_shock = np.random.normal(0, volatility)

            growth = baseline_growth + reconstruction_component + random_shock

        else:
            growth = baseline_growth

        gdp.append(gdp[-1] * (1 + growth))

    return gdp



np.random.seed(42)  # for reproducibility

gdp_low = simulate_recovery(reconstruction_low)
gdp_medium = simulate_recovery(reconstruction_medium)
gdp_high = simulate_recovery(reconstruction_high)



plt.figure(figsize=(10, 6))

plt.plot(gdp_low, label="Low Reconstruction (4%)")
plt.plot(gdp_medium, label="Medium Reconstruction (6%)")
plt.plot(gdp_high, label="High Reconstruction (8%)")

plt.axvline(conflict_year, linestyle="--", alpha=0.6)

plt.title("Post-Conflict GDP Recovery Simulation (With Instability)")
plt.xlabel("Year")
plt.ylabel("GDP Index")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
