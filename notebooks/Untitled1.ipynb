```python
import pandas as pd


def load_worldbank(filepath):
    df = pd.read_csv(filepath, skiprows=4)

    df = df.melt(
        id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
        var_name="Year",
        value_name="GDP Growth"
    )

    df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
    df["GDP Growth"] = pd.to_numeric(df["GDP Growth"], errors="coerce")

    df = df.dropna(subset=["GDP Growth"])
    df = df.sort_values("Year")

    return df



rwanda = load_worldbank("../data/rwanda_gdp_growth.csv")
bosnia = load_worldbank("../data/bosnia_gdp_growth.csv")



def recovery_metrics(df, conflict_year):

    
    post_conflict = df[(df["Year"] >= conflict_year) & 
                   (df["Year"] <= conflict_year + 10)]
    rwanda[(rwanda["Year"] >= 1994) & (rwanda["Year"] <= 1999)]
    bosnia[(bosnia["Year"] >= 1992) & (bosnia["Year"] <= 1999)]


    trough_row = post_conflict.loc[post_conflict["GDP Growth"].idxmin()]
    trough_year = trough_row["Year"]

    
    after_trough = df[df["Year"] > trough_year]
    positive_year = after_trough[after_trough["GDP Growth"] > 0]["Year"].min()

    years_to_positive = positive_year - trough_year

    
    sustained_year = None

    for i in range(len(after_trough) - 2):
        window = after_trough.iloc[i:i+3]
       if (window["GDP Growth"] > 5).all():
            sustained_year = window.iloc[0]["Year"]
            break

    if sustained_year:
        years_to_sustained = sustained_year - trough_year
    else:
        years_to_sustained = None

    return trough_year, years_to_positive, years_to_sustained



rwanda_metrics = recovery_metrics(rwanda, 1994)
bosnia_metrics = recovery_metrics(bosnia, 1992)

print("Rwanda Recovery Metrics:")
print("Trough Year:", rwanda_metrics[0])
print("Years to Positive Growth:", rwanda_metrics[1])
print("Years to Sustained Growth:", rwanda_metrics[2])

print("\nBosnia Recovery Metrics:")
print("Trough Year:", bosnia_metrics[0])
print("Years to Positive Growth:", bosnia_metrics[1])
print("Years to Sustained Growth:", bosnia_metrics[2])
```
